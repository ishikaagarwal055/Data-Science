Task 1--> Student data manager

```python
class Student:
  def __init__(self,roll_no,name,age):
    self.roll_no = roll_no
    self.name = name
    self.age = age

  def display(self):
   print(f"Roll No: {self.roll_no}")
   print(f"Name: {self.name}")
   print(f"Age: {self.age}")

class StudentManager:
  def __init__(self):
    self.students = []

  def add_student(self):
    roll_no = input("Enter roll no: ")
    name = input("Enter name: ")
    age = input("Enter age: ")
    student = Student(roll_no, name, age)
    self.students.append(student)
    print("Student added successfully!")

  def display_students(self):
    for student in self.students:
      student.display()

  def search_student(self, roll_no):
    for student in self.students:
      if student.roll_no == roll_no:
        return student
    return None

  def delete_student(self, roll_no):
    student = self.search_student(roll_no)
    if student is not None:
      self.students.remove(student)

  def update_student(self, roll_no, new_name, new_age):
    student = self.search_student(roll_no)
    if student is not None:
      student.name = new_name
      student.age = new_age
      print("Student updated successfully!")


student_manager = StudentManager()

while True:
  print("1. Add Student")
  print("2. Display Students")
  print("3. Search Student")
  print("4. Delete Student")
  print("5. Update Student")
  print("6. Exit")

  choice = input("Enter your choice: ")

  if choice =="1":
    student_manager.add_student()
  elif choice =="2":
    student_manager.display_students()
  elif choice =="3":
    roll_no = input("Enter roll no to search: ")
    student = student_manager.search_student(roll_no)
    if student is not None:
      print("Student found:")
      student.display()
  elif choice =="4":
    roll_no = input("Enter roll no to delete: ")
    student_manager.delete_student(roll_no)
  elif choice =="5":
    roll_no = input("Enter roll no to update: ")
    new_name = input("Enter new name: ")
    new_age = input("Enter new age: ")
    student_manager.update_student(roll_no, new_name, new_age)
  elif choice =="6":
    print("Thank youu!")
    break
  else:
    print("Invalid choice. Please try again.")

```
