```python
#reshape array 2d
import numpy as np
arr = np.arange(12) #12 =>3*4 ||4*3
up_arr = np.reshape(arr,(3,4)) # 3->row ||4 ->cols
print(up_arr)

#example 2d
import numpy as np
arr = np.arange(18)
up_arr = np.reshape(arr,(9*2))
print(up_arr)

#example 3d
import numpy as np
arr = np.arange(18)
up_arr = np.reshape(arr,(3*3*2))
print(up_arr)
```

```python
#lattin
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)
up_arr = arr.flatten()
up_arr[0] = 100
print(up_arr)
print(arr)

#example 3d
import numpy as np
arr = np.arange(18).reshape(3,3,2)
print(arr)
up_arr = arr.flatten()
up_arr[0] = 100
print(up_arr)
print(arr)
```

```python
#ravel -> works on original array
#example 2d
import numpy as np
arr = np.arange(12).reshape(6,2)
print(arr)
up_arr = arr.ravel()
up_arr[0] = 100
print(up_arr)
print(arr)

#example 3d
import numpy as np
arr = np.arange(18).reshape(3,3,2)
print(arr)
up_arr = arr.ravel()
up_arr[0] = 100
print(up_arr)
print(arr)
```

```python
#transpose -> rows convert into cols || cols convert into rows
#2d
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)
up_arr = arr.transpose()
print(up_arr)

#3d
import numpy as np
arr = np.arange(18).reshape(3,3,2)
print(arr)
up_arr = arr.transpose()
print(up_arr)
```

```python
import numpy as np

# Data Structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("1. Complete Dataset:")
print(sales_data)

print("\n2. Shape of Array:")
print(sales_data.shape)

print("\n3. Number of Dimensions:")
print(sales_data.ndim)

print("\n4. Total Number of Elements:")
print(sales_data.size)

print("\n5. Restaurant IDs:")
print(sales_data[:, 0])

print("\n6. 2021 Sales Data:")
print(sales_data[:, 1])

print("\n7. 2024 Sales Data:")
print(sales_data[:, 4])

print("\n8. Maximum Sales Value:")
print(np.max(sales_data[:, 1:]))

print("\n9. Minimum Sales Value:")
print(np.min(sales_data[:, 1:]))

print("\n10. Total Sales in 2021:")
print(np.sum(sales_data[:, 1]))

print("\n11. Average Sales in 2024:")
print(np.mean(sales_data[:, 4]))

print("\n12. Total Sales of Each Restaurant (2021-2024):")
print(np.sum(sales_data[:, 1:], axis=1))

print("\n13. Average Sales of Each Restaurant:")
print(np.mean(sales_data[:, 1:], axis=1))

print("\n14. Restaurant with Highest 2024 Sales:")
highest_2024_index = np.argmax(sales_data[:, 4])
print(sales_data[highest_2024_index])

print("\n15. Year-wise Total Sales (2021-2024):")
print(np.sum(sales_data[:, 1:], axis=0))

print("\n16. Year-wise Average Sales (2021-2024):")
print(np.mean(sales_data[:, 1:], axis=0))

print("\n17. Add New Column for 2025 Sales:")
sales_2025 = np.array([[280000],
                       [210000],
                       [330000],
                       [300000],
                       [250000]])

updated_data = np.hstack((sales_data, sales_2025))
print(updated_data)

print("\n18. Delete Restaurant ID 3 Data:")
deleted_data = np.delete(sales_data, 2, axis=0)
print(deleted_data)

print("\n19. Growth from 2023 to 2024:")
growth = sales_data[:, 4] - sales_data[:, 3]
print(growth)

print("\n20. Restaurant with Maximum Total Sales:")
total_sales = np.sum(sales_data[:, 1:], axis=1)
max_index = np.argmax(total_sales)

print("Restaurant ID:", sales_data[max_index, 0])
print("Total Sales:", total_sales[max_index])
```

```python
import numpy as np
arr = np.array([10,20,3,4,90,100])
print(arr)
arr_f = arr[[0,2]] # 0 index value, 2 index value
print(arr_f)

import numpy as np
arr=np.array([10,20,3,4,90,100])
print(arr)

arr_f = arr[[0,2,5]]
print(arr_f)
```

```python
#np.where
import numpy as np
#1d
arr = np.array([10,3,4,80,30,40,9])
print(arr)
arr_w = np.where(arr>40,"True","False")
print(arr_w)

#example
arr = np.array([10,3,4,80,30,40,9])
print(arr)
arr_w = np.where(arr>40,arr+2,arr-2) # condition : true :false
#if (arr>40)
# arr+2
#else
# arr-2
print(arr_w)

#2d
arr = np.array([[10,2,34],[32,54,6]])
print(arr)
arr_w = np.where(arr>20,"True","False")
print(arr_w)
```
