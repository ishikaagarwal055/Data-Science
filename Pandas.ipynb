**Row **                     
 loc -> label based           
 iloc -> index based

```python
import pandas as pd
d = {
    "name":["vishal","virat","vineet"],
    "salary":[100,200,300]
}
df = pd.DataFrame(data=d)

print(df.loc[2,"name"])
print(df.iloc[2,0])
print()

#get single row data
print(df.iloc[1])
print()

#get single row using loc
print(df.loc[1])
print()

#get multiple rows using iloc
print(df.iloc[0:2])
print()

#get multiple rows using loc
print(df.loc[0:3])
print()

#sub data get using iloc
df1 = df.iloc[0:2,[0]] # rows->0,1 and cols ->0 |name
df1
print()

#sub data get using loc
df2 = df.loc[0:1,["name"]]
df2
```
