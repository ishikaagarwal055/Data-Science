# E-commerce Data Analysis

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

url = "https://raw.githubusercontent.com/ishikaagarwal055/Data-Science/main/Ecommerce.json"
df = pd.read_json(url)
df

# Rating --> mean
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())
# df.fillna(df["Rating"].mean(),inplace=True)

# Price --> median
df['Price'] = df['Price'].fillna(df["Price"].median())

# Row 8 --> drop
df = df.dropna(subset=['Quantity'])

# deleting the index of droped value
df.reset_index(drop=True, inplace=True)

plt.style.use('ggplot')

# Subplots

fig , aux = plt.subplots(5,3,figsize=(27,30))

aux[0][0].bar(df['Order_ID'],df['Customer'])
aux[0][0].set_title('Order_ID')
aux[0][0].set_xlabel('Number of orders per customer')
aux[0][0].set_ylabel('Customer')

aux[0][1].bar(df['Order_ID'],df['Category'])
aux[0][1].set_title('Order_ID with Category')
aux[0][1].set_xlabel('Order_ID')
aux[0][1].set_ylabel('Category')

aux[0][2].pie(df['Order_ID'],labels=df['Price'], autopct='%1.1f%%')
aux[0][2].set_title('Order_ID with Price')
aux[0][2].set_xlabel('Order_ID')
aux[0][2].set_ylabel('Price')

aux[1][0].plot(df['Order_ID'],df['Quantity'])
aux[1][0].set_title('Order_ID with Quantity')
aux[1][0].set_xlabel('Order_ID')
aux[1][0].set_ylabel('Quantity')

aux[1][1].scatter(df['Order_ID'],df['Product'])
aux[1][1].set_title('Order_ID with Product')
aux[1][1].set_xlabel('Order_ID')
aux[1][1].set_ylabel('Product')

aux[1][2].plot(df['Customer'],df['Category'])
aux[1][2].set_title('Customer with Category')
aux[1][2].set_xlabel('Customer')
aux[1][2].set_ylabel('Category')

aux[2][0].bar(df['Customer'],df['Product'])
aux[2][0].set_title('Customer with product')
aux[2][0].set_xlabel('Customer')
aux[2][0].set_ylabel('Product')

aux[2][1].pie(df['Quantity'],labels=df['Customer'], autopct='%1.1f%%')
aux[2][1].set_title('Quantity with Customer')
aux[2][1].set_xlabel('Quantity')
aux[2][1].set_ylabel('QCustomer')

aux[2][2].scatter(df['Category'],df['Product'])
aux[2][2].set_title('Category with Product')
aux[2][2].set_xlabel('Category')
aux[2][2].set_ylabel('Product')

aux[3][0].bar(df['Category'],df['Quantity'])
aux[3][0].set_title('Category with Quantity')
aux[3][0].set_xlabel('Category')
aux[3][0].set_ylabel('Quantity')

aux[3][1].pie(df['Price'],labels=df['Category'], autopct='%1.1f%%')
aux[3][1].set_title('Price with Category')
aux[3][1].set_xlabel('Price')
aux[3][1].set_ylabel('Category')

aux[3][2].hist(df['Rating'], bins=5, edgecolor='black')
aux[3][2].set_title('Rating with Product')
aux[3][2].set_xlabel('Rating')
aux[3][2].set_ylabel('Product')

aux[4][0].bar(df['Product'],df['Quantity'])
aux[4][0].set_title('Product with Quantity')
aux[4][0].set_xlabel('Product')
aux[4][0].set_ylabel('Quantity')

aux[4][1].pie(df['Price'],labels=df['Product'], autopct='%1.1f%%')
aux[4][1].set_title('Price with Product')
aux[4][1].set_xlabel('Price')
aux[4][1].set_ylabel('Product')

aux[4][2].hist(df['Order_Date'], bins=5, edgecolor='black')
aux[4][2].set_title('Order_Date with Product')
aux[4][2].set_xlabel('Order_Date')
aux[4][2].set_ylabel('Product')

# Saving my file
plt.savefig('E-commerce.png')
plt.tight_layout()
plt.show()
```
